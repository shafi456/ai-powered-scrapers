```python
from openai import OpenAI
from scraper import fetch_website_contents
from IPython.display import Markdown, display, update_display

## Prerequisites
# 1. Install Ollama from https://ollama.com/download (Windows/Mac/Linux)
# 2. Start it — it usually runs automatically after install. If not, run `ollama serve` in a terminal and leave it running.
# 3. Pull the model this project uses:
#    ollama pull qwen3.5:4b


system_prompt = """
You are an elite, no-nonsense competitive programming parser. Your sole purpose is to bypass all narrative lore in coding problems and extract only the cold, 
hard mathematical and algorithmic requirements. Be direct and concise. Format strictly in Markdown.
"""

# Define our user prompt

user_prompt_prefix = """
Analyze the following competitive programming problem. Strip away all story elements, character names, and lore. 

Provide exactly the following:
- **Core Challenge:** 1-2 sentences explaining the exact algorithmic task.
- **Inputs:** The explicit input format and data types.
- **Outputs:** The explicit output format.
- **Constraints:** A concise bulleted list of all time, space, and variable limits.

"""


def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    OLLAMA_BASE_URL = "http://localhost:11434/v1"
    ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
    website = fetch_website_contents(url)
    stream = ollama.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(website),
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    # Print each word the moment the model generates it
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)

# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://leetcode.com/problems/maximum-distance-between-a-pair-of-values/description/")
```
