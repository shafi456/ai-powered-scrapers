```python
from openai import OpenAI
from scraper import fetch_website_contents
from IPython.display import Markdown, display, update_display

## Prerequisites
# 1. Install Ollama from https://ollama.com/download (Windows/Mac/Linux)
# 2. Start it — it usually runs automatically after install. If not, run `ollama serve` in a terminal and leave it running.
# 3. Pull the model this project uses:
#    ollama pull llama3.2


system_prompt = """
You are a hyper-efficient gaming assistant focused strictly on optimizing in-game currency and rewards. 
You ignore game lore, storyline dialogue, and promotional filler. Output strictly formatted, actionable checklists.
"""

# Define our user prompt

user_prompt_prefix = """
Analyze the following Genshin Impact event or patch notes. Ignore all character dialogue and story summaries. 

Extract the following details:
- **Total Primogems:** The absolute maximum number of Primogems available.
- **Prerequisites:** Any required Adventure Rank, previous quests, or energy thresholds to participate.
- **Actionable Checklist:** A bulleted to-do list of the exact actions required to claim all rewards and character ascension materials.
- **Deadlines:** The end dates for the events.

"""


def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    OLLAMA_BASE_URL = "http://localhost:11434/v1"
    ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
    website = fetch_website_contents(url)
    stream = ollama.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(website),
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    # Print each word the moment the model generates it
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)

# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://genshin.hoyoverse.com/en/news/detail/164999")
```
