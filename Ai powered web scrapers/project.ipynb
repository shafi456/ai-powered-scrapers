```python
from openai import OpenAI
from scraper import fetch_website_contents
from IPython.display import Markdown, display

## Prerequisites
# 1. Install Ollama from https://ollama.com/download (Windows/Mac/Linux)
# 2. Start it — it usually runs automatically after install. If not, run `ollama serve` in a terminal and leave it running.
# 3. Pull the model this project uses:
#    ollama pull deepseek-r1:1.5b


system_prompt = """
You are a ruthlessly pragmatic financial advisor and tech critic. Your goal is to aggressively roast e-commerce products and stop the user from making impulse purchases. 
Ignore all marketing hype. Be witty, blunt, and highly sarcastic. Format your response cleanly in Markdown without any introductory pleasantries.
"""

# Define our user prompt

user_prompt_prefix = """
Analyze the following product page content. 

1. Extract a bulleted list of the actual technical specifications (ignore buzzwords).
2. Roast the marketing hype.
3. Provide three brutal, logical reasons why I do not need to buy this right now.

"""


def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    OLLAMA_BASE_URL = "http://localhost:11434/v1"
    ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model = "deepseek-r1:1.5b",
        messages = messages_for(website),
        stream=True
    )
    full_response = ""
    # Print each word the moment the model generates it
    for chunk in response:
        content = chunk.choices[0].delta.content
        if content:
            print(content, end="", flush=True)
            full_response += content
            
    return full_response

# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://www.startech.com.bd/lenovo-ideapad-1-15amn7-ryzen-5-laptop")
```
